# 02 — Series

## Code

```python
data = [10, 20, 30, 40]

s = pd.Series(data)

s
```

---

## Explanation

* `Series` Pandas ka **one-dimensional data structure** hai.
* Series me data ke saath index bhi hota hai.
* Default index `0` se start hota hai.

---

## Example

```python
data = [10, 20, 30]

s = pd.Series(data)

s
```

### Output

```text
0    10
1    20
2    30
dtype: int64
```

---

## Custom Index

```python
s = pd.Series(
    [10, 20, 30],
    index=["A", "B", "C"]
)

s
```

---

## ⭐ Interview Important

# 46 — shift(freq="D")

## Code

```python
df.shift(freq="D")
```

---

## Explanation

* `shift(freq="D")` DatetimeIndex ko 1 day shift karta hai.
* Data/Values same rehte hain.
* Sirf Date/Time Index change hota hai.

---

## Different Frequencies

```python
df.shift(freq="H")      # 1 Hour

df.shift(freq="D")      # 1 Day

df.shift(freq="W")      # 1 Week

df.shift(freq="ME")     # Month End

df.shift(freq="YE")     # Year End
```

---

## ⭐ Interview Important

```python
# IMPORTANT:

# shift(freq="D")
# -> DatetimeIndex ko 1 Day shift karta hai.

# Data change nahi hota.

# Sirf Date/Time Index shift hota hai.

# freq="D" -> Daily Frequency
# freq="H" -> Hourly Frequency
# freq="W" -> Weekly Frequency
# freq="ME" -> Month End
# freq="YE" -> Year End
```

---

## Interview Questions & Answers

### Q1. `shift(freq="D")` kya karta hai?

**Answer:**

DatetimeIndex ko 1 day aage shift karta hai.

### Q2. Kya `shift(freq="D")` data ko change karta hai?

**Answer:**

Nahi. Sirf Date/Time Index change hota hai.

### Q3. `shift(1)` aur `shift(freq="D")` me difference kya hai?

**Answer:**

* `shift(1)` → Data/Values ko 1 row shift karta hai.
* `shift(freq="D")` → DatetimeIndex ko 1 day shift karta hai.
