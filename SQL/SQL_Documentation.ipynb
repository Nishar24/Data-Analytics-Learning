# 69 — JOINS

SQL ka **very important topic**.

Main JOIN types:

```text
INNER JOIN
LEFT JOIN
RIGHT JOIN
FULL OUTER JOIN
CROSS JOIN
SELF JOIN
```

---

# 70 — INNER JOIN

Sirf matching records:

```sql
SELECT
    customers.name,
    orders.order_id
FROM customers
INNER JOIN orders
ON customers.customer_id =
   orders.customer_id;
```

Concept:

```text
Table A
   +
Table B
   ↓
Matching records
```

---

# 71 — LEFT JOIN

Left table ke **saare records** + matching right table records.

```sql
SELECT
    customers.name,
    orders.order_id
FROM customers
LEFT JOIN orders
ON customers.customer_id =
   orders.customer_id;
```

Agar order nahi hai to order columns me `NULL` aa sakta hai.

---

# 72 — RIGHT JOIN

Right table ke saare records + matching left table records.

```sql
SELECT
    customers.name,
    orders.order_id
FROM customers
RIGHT JOIN orders
ON customers.customer_id =
   orders.customer_id;
```

---

# 73 — FULL OUTER JOIN

Dono tables ke all records, matching ho ya na ho.

Concept:

```text
LEFT records
+
Matching records
+
RIGHT-only records
```

> MySQL me `FULL OUTER JOIN` directly supported nahi hota; commonly `LEFT JOIN` + `RIGHT JOIN` ko `UNION` ke through emulate kiya jata hai.

# 95 — SQL Query Execution Order

Ye **interview ke liye bahut important** hai.

Query:

```sql
SELECT department,
       AVG(salary) AS avg_salary
FROM employees
WHERE salary > 30000
GROUP BY department
HAVING AVG(salary) > 50000
ORDER BY avg_salary DESC
LIMIT 5;
```

Logical execution order roughly:

```text
FROM
↓
WHERE
↓
GROUP BY
↓
HAVING
↓
SELECT
↓
ORDER BY
↓
LIMIT
```

> SQL ko hum generally `SELECT` se likhte hain, lekin logical processing order alag hota hai.

# 96 — WHERE vs HAVING

| WHERE                                                           | HAVING                              |
| --------------------------------------------------------------- | ----------------------------------- |
| Rows filter karta hai                                           | Groups filter karta hai             |
| GROUP BY se pehle logically apply hota hai                      | GROUP BY ke baad                    |
| Aggregate result ko directly filter karne ke liye normally nahi | Aggregate conditions ke liye useful |

Example:

```sql
WHERE salary > 50000
```

vs

```sql
HAVING AVG(salary) > 50000
```

# 114 — Top N per Group

Example: har department ke top 3 employees.

```sql
WITH ranked_employees AS (

    SELECT
        name,
        department,
        salary,

        ROW_NUMBER() OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS rn

    FROM employees

)

SELECT *
FROM ranked_employees
WHERE rn <= 3;
```

# 115 — Second Highest Salary

Simple approach:

```sql
SELECT MAX(salary) AS second_highest
FROM employees
WHERE salary < (
    SELECT MAX(salary)
    FROM employees
);
```

Window function approach:

```sql
WITH ranked AS (

    SELECT
        salary,
        DENSE_RANK() OVER (
            ORDER BY salary DESC
        ) AS rnk
    FROM employees

)

SELECT salary
FROM ranked
WHERE rnk = 2;
```

# 116 — Duplicate Records Find Karna

Example:

```sql
SELECT
    email,
    COUNT(*) AS total
FROM users
GROUP BY email
HAVING COUNT(*) > 1;
```

Duplicate emails mil jayenge.

# 117 — Duplicate Records Remove Karna

Actual deletion se pehle duplicate identification aur backup/testing zaroor karo.

Window-function based approach:

```sql
WITH duplicates AS (

    SELECT
        id,
        ROW_NUMBER() OVER (
            PARTITION BY email
            ORDER BY id
        ) AS rn
    FROM users

)

DELETE FROM users
WHERE id IN (
    SELECT id
    FROM duplicates
    WHERE rn > 1
);
```

> Exact syntax database engine ke according vary kar sakti hai; production me deletion se pehle query ko carefully test karna chahiye.

# 118 — SQL Interview Questions

### Q1. SQL kya hai?

**Answer:**
SQL ek language hai jo relational databases me data ko query aur manipulate karne ke liye use hoti hai.

### Q2. Primary Key kya hai?

**Answer:**
Primary Key table ke har record ko uniquely identify karti hai aur NULL nahi hoti.

### Q3. Foreign Key kya hai?

**Answer:**
Foreign Key ek table ke column ko doosri table ke candidate/primary key se relate karti hai.

### Q4. WHERE aur HAVING me difference?

**Answer:**
`WHERE` rows ko filter karta hai, jabki `HAVING` grouped/aggregated results ko filter karne ke liye use hota hai.

### Q5. DELETE, TRUNCATE aur DROP me difference?

**Answer:**

```text
DELETE
→ Rows delete

TRUNCATE
→ All rows remove

DROP
→ Table structure + data remove
```

### Q6. INNER JOIN kya karta hai?

**Answer:**
Dono tables ke matching records return karta hai.

### Q7. LEFT JOIN kya karta hai?

**Answer:**
Left table ke saare records aur right table ke matching records return karta hai.

### Q8. UNION vs UNION ALL?

**Answer:**
`UNION` duplicate rows remove karta hai, jabki `UNION ALL` duplicates ko retain karta hai.

### Q9. GROUP BY kya karta hai?

**Answer:**
Same values ko groups me divide karta hai aur aggregate functions ke saath commonly use hota hai.

### Q10. Index kya hai?

**Answer:**
Index query performance improve karne ke liye use hone wala data structure hai, lekin additional storage aur write overhead bhi ho sakta hai.

### Q11. Subquery kya hai?

**Answer:**
Ek SQL query ke andar execute hone wali doosri query ko subquery kehte hain.

### Q12. CTE kya hai?

**Answer:**
CTE ek temporary named result set hai jo `WITH` clause se define hota hai aur complex queries ko readable banata hai.

### Q13. Window Function kya hai?

**Answer:**
Window function rows ko collapse kiye bina related rows ke set par calculation perform karta hai.

### Q14. RANK aur DENSE_RANK me difference?

**Answer:**
`RANK()` ties ke baad rank gaps create kar sakta hai, jabki `DENSE_RANK()` gaps nahi create karta.

### Q15. SQL query ka logical execution order?

**Answer:**

```text
FROM
↓
WHERE
↓
GROUP BY
↓
HAVING
↓
SELECT
↓
ORDER BY
↓
LIMIT
```
