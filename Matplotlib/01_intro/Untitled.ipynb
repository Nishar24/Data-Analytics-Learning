## 1. Matplotlib kya hai?

**Matplotlib** Python ki ek popular **Data Visualization Library** hai.

Iska use data ko different types ke graphs/charts ke form me represent karne ke liye hota hai.

### Matplotlib se hum bana sakte hain:

* Line Plot
* Bar Chart
* Histogram
* Pie Chart
* Scatter Plot
* Subplots
* etc.

### Data Analyst ke liye importance

Raw data ko dekhkar trend samajhna difficult ho sakta hai.

Graph ki help se hum easily:

* Trend dekh sakte hain
* Comparison kar sakte hain
* Pattern identify kar sakte hain
* Outliers dekh sakte hain
* Data ko visually present kar sakte hain

# 3. Basic Line Plot

Basic syntax:

```python
plt.plot(x, y)
```

Yahan:

* `x` → X-axis ka data
* `y` → Y-axis ka data

Example:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [20, 25, 30, 28, 35]

plt.plot(x, y)
plt.show()
```

Ye X aur Y values ke according ek **line graph** create karega.

---

# 10. Complete Example

Ab sab parameters ko ek saath use karte hain:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [20, 25, 30, 28, 35]

plt.xlabel('Day')
plt.ylabel('Temperature')
plt.title('Weather')

plt.plot(
    x,
    y,
    color='green',
    linewidth=5,
    linestyle='dotted'
)

plt.show()
```

### Is graph me:

* X-axis → `Day`
* Y-axis → `Temperature`
* Title → `Weather`
* Line color → Green
* Line thickness → 5
* Line style → Dotted

# 13. Real-Life Data Analyst Example

Suppose ek company ke paas 5 days ka temperature data hai:

```python
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 33]
```

Hum iska trend visualize kar sakte hain:

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 33]

plt.xlabel('Day')
plt.ylabel('Temperature')
plt.title('Temperature Trend')

plt.plot(
    days,
    temperature,
    color='green',
    linewidth=3,
    linestyle='dotted'
)

plt.show()
```

Isse analyst easily dekh sakta hai ki temperature kis direction me move kar raha hai.
