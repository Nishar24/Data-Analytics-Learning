# 05 — Histogram

# 1. Basic Histogram

## Code

```python
blood_sugar = [113, 85, 90, 150, 149, 88, 93, 115, 135, 80, 77, 82, 129]

plt.hist(blood_sugar)
```

## Explanation

- `plt.hist()` → Histogram create karta hai.
- Histogram ka use numerical data ki **distribution** dekhne ke liye hota hai.
- Data ko different ranges/groups me divide kiya jata hai.
- In ranges ko **bins** kaha jata hai.
- Har bin me kitni values hain, uski frequency bar ki height se show hoti hai.

### Simple Concept

```text
Raw Data
   ↓
Ranges / Bins
   ↓
Har range me values ka count
   ↓
Histogram
   ↓
Data Distribution
```

---

# 2. `bins`

## Code

```python
plt.hist(blood_sugar, bins=3)
```

## Explanation

- `bins` → Data ko kitne groups/ranges me divide karna hai, ye decide karta hai.
- `bins=3` → Data ko 3 bins me divide karega.
- Bins ki value increase karne par distribution zyada detailed dikhegi.
- Bins ki value kam karne par distribution broader dikhegi.

---

# 3. `rwidth`

## Code

```python
plt.hist(
    blood_sugar,
    bins=3,
    rwidth=0.95
)
```

## Explanation

- `rwidth` → Histogram bars ki **relative width** control karta hai.
- `rwidth=0.95` → Bar available bin width ka approximately 95% use karega.
- Isse bars ke beech halka gap दिखाई deta hai.

### Examples

```text
rwidth = 1.0   → Full/almost full width
rwidth = 0.95  → Thoda gap
rwidth = 0.5   → Zyada gap
```

---

# 4. Custom Bins

## Code

```python
plt.hist(
    blood_sugar,
    bins=[80, 100, 125, 150],
    rwidth=0.95
)
```

## Explanation

Yahan humne bins ko manually define kiya hai.

```text
80–100
100–125
125–150
```

### Important Rule

```text
Bin boundaries ki count - 1 = Number of bins
```

Example:

```text
[80, 100, 125, 150]

4 boundaries
     ↓
3 bins
```

### Blood Sugar Distribution

| Range | Classification |
|---|---|
| 80–100 | Normal |
| 100–125 | Pre-Diabetic |
| 125–150 | Diabetic |

> Note: Ye classification sirf lecture/example ke context me use ki gayi ranges hain; real medical diagnosis ke liye clinical criteria follow kiye jaate hain.

### 2. Histogram vs Bar Chart

```text
Histogram
    ↓
Numerical data ki distribution
    ↓
Data ranges/bins me divide hota hai

Bar Chart
    ↓
Categories ki values compare karta hai
    ↓
Har category ka separate bar
```

# Interview Questions & Answers

### Q1. `plt.hist()` kya karta hai?

**Answer:**

`plt.hist()` data ka Histogram create karta hai, jisse numerical data ki distribution aur frequency samajh sakte hain.

---

### Q2. Histogram ka main purpose kya hai?

**Answer:**

Numerical data ki **distribution** dekhna aur ye samajhna ki values kis range me zyada concentrated hain.

---

### Q3. Histogram me bins kya hote hain?

**Answer:**

Bins data ko divide kiye gaye **ranges/intervals** hote hain.

---

### Q4. `bins=3` ka kya matlab hai?

**Answer:**

Data ko 3 bins/ranges me divide karna.

---

### Q5. `bins=[80,100,125,150]` se kitne bins banenge?

**Answer:**

3 bins.

```text
80–100
100–125
125–150
```

Kyuki:

```text
4 boundaries - 1 = 3 bins
```

---

### Q6. `rwidth` kya karta hai?

**Answer:**

Histogram bars ki relative width control karta hai.

---

### Q7. `rwidth=0.95` ka kya matlab hai?

**Answer:**

Bars available bin width ka approximately 95% space occupy karengi.

---

### Q8. `histtype='step'` kya karta hai?

**Answer:**

Histogram ko filled bars ki jagah **outline/step form** me display karta hai.

---

### Q9. Kya `plt.hist()` multiple datasets ke liye use ho sakta hai?

**Answer:**

Haan.

```python

```

**Answer:**

`color` me colors ki list provide karenge.

### Q13. `plt.grid()` kya karta hai?

**Answer:**

Graph par grid lines display karta hai, jisse values ko read karna easier hota hai.
