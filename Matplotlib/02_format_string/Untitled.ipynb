## 11. Format String

Matplotlib me **Format String** ek short notation hai jisse `plot()` ke andar:

* Color
* Marker
* Line Style

ko ek saath specify kar sakte hain.

### Syntax

```python
plt.plot(x, y, 'format_string')
```

### Example

```python
plt.plot(x, y, 'g+--')
```

Yahan `'g+--'` ek **Format String** hai.

```text
'g+--'
 │││
 ││└── Line Style
 │└─── Marker
 └──── Color
```

### Meaning

* `g` → Green color
* `+` → Plus marker
* `--` → Dashed line

Isliye:

```python
plt.plot(x, y, 'g+--')
```

ka matlab hai:

> Green color ki dashed line aur har data point par `+` marker.

---

## 12. Format String Components

Format String mainly 3 components se milkar bana hota hai:

```text
Color + Marker + Line Style
```

Example:

```python
plt.plot(x, y, 'g+--')
```

```text
g   → Color
+   → Marker
--  → Line Style
```

### 12.3 Line Style Codes

| Code | Line Style |

## 13. Format String Examples

```python
plt.plot(x, y, 'r--')      # Red + Dashed

```

```text
 │└─── Star marker
 └──── Red
```


## 14. `alpha` Parameter

`alpha` parameter ka use plot ki **transparency/opacity** control karne ke liye hota hai.

### Example

```python
plt.plot(x, y, color='green', alpha=0.2)
```

Yahan:

```python
color='green'
```

→ Line ka color green hai.

```python
alpha=0.2
```

→ Line ko transparent banata hai.

---

## 15. Alpha Values

`alpha` ki value generally `0` se `1` ke beech hoti hai.

| Alpha | Meaning                |
| ----: | ---------------------- |
|   `0` | Completely transparent |
| `0.2` | Highly transparent     |
| `0.5` | Medium transparency    |
| `0.8` | Slightly transparent   |
|   `1` | Fully visible / opaque |

### Examples

```python
plt.plot(x, y, color='green', alpha=0.2)
```

→ Bahut transparent line.

```python
plt.plot(x, y, color='green', alpha=0.5)
```

→ Medium transparency.

```python
plt.plot(x, y, color='green', alpha=1)
```

→ Fully visible line.

### 🔥 Important Point

> `alpha` ka main purpose **transparency control karna** hai.

### 🎯 Interview Question

**Q. Matplotlib me `alpha` parameter kya karta hai?**

**Answer:**
`alpha` plot ki transparency/opacity control karta hai. Iski value generally `0` se `1` ke beech hoti hai.

# Seaborn

## 16. Seaborn Kya Hai?

**Seaborn** Python ki ek **statistical data visualization library** hai.

Seaborn **Matplotlib ke upar built** hai aur attractive statistical visualizations banana easy banata hai.

### Import

```python
import seaborn as sns
```

Yahan:

* `seaborn` → Library
* `sns` → Seaborn ka commonly used alias

---

## 17. `sns.set_theme()`

```python
sns.set_theme(style="darkgrid")
```

`set_theme()` Seaborn plots ka **overall visual theme/style** set karta hai.

### Example

```python
import seaborn as sns

sns.set_theme(style="darkgrid")
```

### Important Point

`sns.set_theme()` **data ko change nahi karta**.

Ye sirf graph ki **appearance/style** change karta hai.

### 🎯 Interview Question

**Q. `sns.set_theme()` ka use kya hai?**

**Answer:**
Seaborn plots ka overall visual theme/style set karne ke liye.

# 20. `pairplot()`

```python
sns.pairplot(df, hue="species")
```

`pairplot()` DataFrame ke multiple numeric columns ke **pair-wise relationships** ko visualize karta hai.

### Example

Agar DataFrame me:

```text
A
B
C
D
```

numeric columns hain, to pairplot relationships jaise:

```text
A ↔ B
A ↔ C
A ↔ D
B ↔ C
B ↔ D
C ↔ D
```

ko visualize karta hai.

### 🔥 Important Point

> `pairplot()` multiple variables ke relationships ko ek hi visualization me compare karne ke liye useful hai.

## 21. `hue` Parameter

```python
sns.pairplot(df, hue="species")
```

Yahan:

```python
hue="species"
```

ka use `species` ke basis par data ko **different groups me distinguish** karne ke liye kiya gaya hai.

### Simple Meaning

> `hue` kisi categorical column ke according observations ko alag-alag groups me show karta hai.

### Example

```python
sns.pairplot(df, hue="species")
```

Yahan different `species` ko alag visual representation milega.

### 🔥 Important Point

`hue` Seaborn ka bahut important parameter hai.

### 🎯 Interview Question

**Q. Seaborn me `hue` parameter ka kya use hai?**

**Answer:**
Categorical variable ke basis par data points/observations ko different groups me distinguish karne ke liye.

---

## 22. `pairplot()` ka Structure

`pairplot()` me generally:

### Diagonal

Diagonal par individual variables ka **distribution** show hota hai.

### Diagonal ke Bahar

Diagonal ke bahar different variables ke beech **relationship/scatter plots** show hote hain.

### Simple Structure

```text
        A       B       C

A       ●       •       •

B       •       ●       •

C       •       •       ●
```

* `●` → Individual variable ka distribution
* `•` → Variables ke beech relationship

## 23. Tumhare Data ke Saath Seaborn

Tumhara data:

```python
x = [1, 2, 3, 4, 5, 6]
y = [50, 51, 48, 47, 49, 46]
```

Isko Seaborn me line plot ke through visualize kar sakte hain:

```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [50, 51, 48, 47, 49, 46]

sns.set_theme(style="darkgrid")

sns.lineplot(x=x, y=y)

plt.show()
```

### 🔥 Important Point

Tumhare data me sirf **2 variables** hain:

* `x`
* `y`

Isliye `lineplot()` yahan `pairplot()` se zyada suitable hai.

`pairplot()` ka main use multiple variables wale DataFrame me hota hai.
