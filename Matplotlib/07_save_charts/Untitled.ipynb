# ⭐ Interview Questions & Answers

### Q1. Matplotlib me chart ko save karne ke liye kya use karte hain?

**Answer:**

```python
plt.savefig()
```

---

### Q2. `plt.savefig("piechart.png")` kya karta hai?

**Answer:**

Current Matplotlib figure ko `piechart.png` naam ki PNG image file ke roop me save karta hai.

---

### Q3. Kya `savefig()` sirf PNG format me save karta hai?

**Answer:**

Nahi.

PNG ke alawa PDF, JPEG, SVG aur other supported formats me bhi save kar sakte hain.

Example:

```python
plt.savefig("chart.pdf")
plt.savefig("chart.jpg")
plt.savefig("chart.svg")
```

---

### Q4. `bbox_inches="tight"` ka use kya hai?

**Answer:**

Figure ke around unnecessary whitespace ko reduce karta hai aur chart ko tightly fit karke save karta hai.

---

### Q5. `pad_inches` kya karta hai?

**Answer:**

Saved figure ke around extra padding/margin control karta hai.

---

### Q6. `transparent=True` kya karta hai?

**Answer:**

Saved figure ka background transparent karta hai.

---

### Q7. `plt.show()` aur `plt.savefig()` me kya difference hai?

**Answer:**

```text
plt.show()
→ Chart ko display karta hai.

plt.savefig()
→ Chart ko file me save karta hai.
```

---

### Q8. `plt.axis("equal")` ka use kya hai?

**Answer:**

X-axis aur Y-axis ka equal scaling maintain karta hai, jisse Pie Chart proper circular shape me appear hota hai.

---

### Q9. `startangle` aur `axis("equal")` me kya difference hai?

**Answer:**

```text
axis("equal")
→ Chart ki shape ko circular maintain karta hai.

startangle
→ Chart ki starting rotation/position change karta hai.
```

---

### Q10. Windows path me `r` ka use kyun karte hain?

**Answer:**

`r` raw string banata hai, jisse backslash `\` ko escape character ki tarah interpret nahi kiya jata.

Example:

```python
r"C:\Users\mdnis\Desktop\piechart.pdf"
```

---

### Q11. `PermissionError: [Errno 13]` kyun aata hai?

**Answer:**

Jab Python ko specified location par file create/write karne ki permission nahi milti, tab `PermissionError` aa sakta hai.

---

### Q12. `savefig()` ko `show()` se pehle kyun use karna better hai?

**Answer:**

Pehle figure ko save karne se intended/current figure reliably save hoti hai.

Recommended:

```python
plt.savefig("piechart.png")
plt.show()
```

---

### Q13. `transparent=True` kab useful hai?

**Answer:**

Jab saved chart ko kisi presentation, website, design ya kisi existing background ke upar use karna ho aur chart ka background transparent rakhna ho.
