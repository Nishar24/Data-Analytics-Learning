````python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
````

### Simple bar chart showing revenues of major US tech companies

````python
company=['GOOGL','AMZN','MSFT','FB']
revenue=[90,136,89,27]
profit=[40,2,34,12]
````

````python
ypos=np.arange(len(company))
ypos

# IMPORTANT:

# company -> Companies ke names ki list.

# revenue -> Har company ki corresponding revenue.

# len(company) -> Companies ki total count.

# np.arange(len(company)) -> 0 se start hone wale positions create karta hai.

# ypos -> Companies ke positions store karta hai.

# Yahan 4 companies hain, isliye ypos = [0, 1, 2, 3].

# ypos ka use Bar Chart me companies ki position
# define karne ke liye kiya jayega.
````

````python
plt.bar(ypos, revenue)
````

````python
plt.xticks(ypos, company)
plt.ylabel("revenue(bln)")
plt.title("US Tech Stocks")
plt.bar(ypos, revenue, label="Revenue")
plt.legend()
````

### Multiple Bars showing revenue and profit of major US tech companies

````python
plt.xticks(ypos, company)
plt.ylabel("revenue(bln)")
plt.title("US Tech Stocks")
plt.bar(ypos, revenue, label="Revenue")
plt.bar(ypos, profit,label="Profit")
plt.legend()
````

````python
plt.xticks(ypos, company)
plt.ylabel("revenue(bln)")
plt.title("US Tech Stocks")

plt.bar(ypos - 0.2, revenue, label="Revenue")
plt.bar(ypos + 0.2, profit, label="Profit")

plt.legend()
plt.show()
````

````python
plt.xticks(ypos, company)
plt.ylabel("revenue(bln)")
plt.title("US Tech Stocks")

plt.bar(ypos - 0.2, revenue, width=0.4, label="Revenue")
plt.bar(ypos + 0.2, profit, width=0.4, label="Profit")

plt.legend()
plt.show()
````

````python
# IMPORTANT:

# width=0.4 -> Bar ki width set karta hai.

# ypos - 0.2 -> Revenue bar ko left shift karta hai.

# ypos + 0.2 -> Profit bar ko right shift karta hai.

# Dono bars ki width 0.4 hai.

# -0.2 aur +0.2 ki wajah se bars side-by-side aate hain.

# Same position use karne par bars overlap kar sakti hain.

# label="Revenue" -> Revenue bar ko naam deta hai.

# label="Profit" -> Profit bar ko naam deta hai.

# plt.legend() -> Dono labels ko graph par show karta hai.

# plt.show() -> Graph ko display karta hai.
````

### Horizontal bar chart using barh function

````python
plt.yticks(ypos, company)
plt.title("US Tech Stocks")

plt.barh(ypos-0.2, revenue, label="Revenue")
````

````python
plt.yticks(ypos, company)
plt.title("US Tech Stocks")

plt.barh(ypos-0.2, revenue, label="Revenue")
plt.barh(ypos+0.2, profit, label="Profit")
plt.legend()
````

````python
# IMPORTANT:

# plt.barh() -> Horizontal Bar Chart banata hai.

# plt.yticks(ypos, company)
# -> Y-axis par company names show karta hai.

# ypos - 0.2 -> Revenue bar ko upar shift karta hai.

# ypos + 0.2 -> Profit bar ko neeche shift karta hai.

# Same ypos use karne par horizontal bars overlap kar sakte hain.

# label="Revenue" -> Revenue ko label deta hai.

# label="Profit" -> Profit ko label deta hai.

# plt.legend() -> Revenue aur Profit labels ko graph par show karta hai.

# Vertical Bar:
# plt.bar() + plt.xticks()

# Horizontal Bar:
# plt.barh() + plt.yticks()
````

# 5. Revenue aur Profit — Grouped Bar Chart

## Code

```python
plt.xticks(ypos, company)
plt.ylabel("revenue(bln)")
plt.title("US Tech Stocks")

plt.bar(ypos - 0.2, revenue, width=0.4, label="Revenue")
plt.bar(ypos + 0.2, profit, width=0.4, label="Profit")

plt.legend()
plt.show()
```

## Explanation

### `plt.bar(ypos - 0.2, revenue, ...)`

```python
plt.bar(ypos - 0.2, revenue, width=0.4, label="Revenue")
```

- `ypos - 0.2` → Revenue bar ko left shift karta hai.
- `revenue` → Revenue ki value/height.
- `width=0.4` → Bar ki width.
- `label="Revenue"` → Bar series ka naam.

### `plt.bar(ypos + 0.2, profit, ...)`

```python
plt.bar(ypos + 0.2, profit, width=0.4, label="Profit")
```

- `ypos + 0.2` → Profit bar ko right shift karta hai.
- `profit` → Profit ki value/height.
- `width=0.4` → Bar ki width.
- `label="Profit"` → Bar series ka naam.

### Side-by-side bars

```text
ypos - 0.2 -> Revenue -> Left
ypos       -> Center
ypos + 0.2 -> Profit  -> Right
```

Isliye Revenue aur Profit ek company ke liye **side-by-side** show hote hain.

# Interview Questions & Answers

### Q1. `plt.bar()` kya karta hai?

**Answer:**

`plt.bar()` ka use Vertical Bar Chart create karne ke liye hota hai.

---

### Q2. `plt.barh()` kya karta hai?

**Answer:**

`plt.barh()` ka use Horizontal Bar Chart create karne ke liye hota hai.

---

### Q3. `plt.bar()` aur `plt.barh()` me kya difference hai?

**Answer:**

- `plt.bar()` → Vertical bars.
- `plt.barh()` → Horizontal bars.

---

### Q4. `plt.xticks(ypos, company)` kya karta hai?

**Answer:**

X-axis ki given positions par company names set karta hai.

---

### Q5. `plt.yticks(ypos, company)` kya karta hai?

**Answer:**

Y-axis ki given positions par company names set karta hai.

---

### Q6. `np.arange()` kya karta hai?

**Answer:**

Given range ke numeric values ki NumPy array create karta hai.

Example:

```python
np.arange(4)
```

Output:

```python
array([0, 1, 2, 3])
```

---

### Q7. `width=0.4` kya karta hai?

**Answer:**

Bar ki width/thickness set karta hai.

---

### Q8. `ypos - 0.2` aur `ypos + 0.2` kyun use kiya?

**Answer:**

Revenue aur Profit bars ko alag positions par shift karke **grouped/side-by-side bar chart** create karne ke liye.

---

### Q9. Agar dono bars me same `ypos` use karein to kya hoga?

**Answer:**

Dono bars same position par draw hongi aur ek doosre ko overlap/cover kar sakti hain.

---

### Q10. `label` parameter ka kya use hai?

**Answer:**

Bar series ko ek naam assign karne ke liye use hota hai.

---

### Q11. `plt.legend()` ka kya use hai?

**Answer:**

`label` ke through diye gaye series names ko graph par display karta hai.

---

### Q12. `plt.show()` kya karta hai?

**Answer:**

Matplotlib ka final graph display karta hai.
