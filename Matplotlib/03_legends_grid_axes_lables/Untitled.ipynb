# 03 — Legends, Grid & Axes Labels

## `legend()` — Graph me Labels Show Karna

---

# Code

```python
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weather")

plt.plot(days, max_t, label="Max")
plt.plot(days, min_t, label="Min")
plt.plot(days, avg_t, label="Avg")

plt.legend()
```

---

# Explanation

* `plt.xlabel("Days")` → X-axis ka label **Days** set karta hai.
* `plt.ylabel("Temperature")` → Y-axis ka label **Temperature** set karta hai.
* `plt.title("Weather")` → Graph ka title **Weather** set karta hai.
* `label="Max"` → Maximum temperature wali line ko **Max** naam deta hai.
* `label="Min"` → Minimum temperature wali line ko **Min** naam deta hai.
* `label="Avg"` → Average temperature wali line ko **Avg** naam deta hai.
* `plt.legend()` → Graph par `Max`, `Min`, aur `Avg` labels ko display karta hai.

---

# Example

```python
days = [1, 2, 3, 4, 5]

max_t = [40, 42, 41, 43, 44]
min_t = [25, 26, 24, 27, 28]
avg_t = [32, 34, 33, 35, 36]

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weather")

plt.plot(days, max_t, label="Max")
plt.plot(days, min_t, label="Min")
plt.plot(days, avg_t, label="Avg")

plt.legend()

plt.show()
```

### Output me

Graph me 3 lines hongi:

* **Max** → Maximum Temperature
* **Min** → Minimum Temperature
* **Avg** → Average Temperature

`plt.legend()` in teeno lines ko identify karne ke liye graph par unke labels show karega.

# Interview Questions & Answers

### Q1. `plt.legend()` kya karta hai?

**Answer:**

Graph me different plots/lines ke labels ko display karta hai.

---

### Q2. `label` parameter ka use kya hai?

**Answer:**

Plot/line ko ek descriptive name assign karne ke liye.

---

### Q3. Kya sirf `label="Max"` likhne se label graph par show ho jayega?

**Answer:**

Nahi.

```python
plt.legend()
```

use karna bhi zaroori hai.

---

### Q4. `label` aur `legend()` me kya difference hai?

**Answer:**

* `label` → Plot ko naam assign karta hai.
* `legend()` → Assigned naam ko graph par display karta hai.

# 💡 Interview Tip

## `label` + `legend()` ka combination yaad rakho

```python
plt.plot(days, max_t, label="Max")
plt.plot(days, min_t, label="Min")
plt.plot(days, avg_t, label="Avg")

plt.legend()
```

Yaad rakho:

| Code           | Kaam                             |
| -------------- | -------------------------------- |
| `label`        | Plot ko naam assign karta hai    |
| `plt.legend()` | Naam graph par display karta hai |

### Short Trick

```text
label  → Naam assign
legend → Naam display
```
