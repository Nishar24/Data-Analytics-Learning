````python
exp_vals = [1400,600,300,410,250]
exp_labels = ["Home Rent", "Food", "Phone/Internet Bill", "Car", "Other Utilities"]
````

# 1. Pie Chart Introduction

Pie Chart ka use kisi total data ko **different categories/parts** me visualize karne ke liye hota hai.

Example:

- Home Rent
- Food
- Phone/Internet Bill
- Car
- Other Utilities

Har category total ka ek portion represent karti hai.

```python
exp_vals = [1400, 600, 300, 410, 250]

exp_labels = [
    "Home Rent",
    "Food",
    "Phone/Internet Bill",

]
```

## Explanation

### `exp_vals`

```python
exp_vals = [1400, 600, 300, 410, 250]
```

| Expense | Amount |
|---|---:|

```python
exp_labels = [

    "Car",
    "Other Utilities"
]
```

- Ye Pie Chart ki categories/labels hain.
- `exp_vals` aur `exp_labels` ka order same hona chahiye.

```text
1400 → Home Rent
600  → Food

410  → Car
250  → Other Utilities
```

# 3. Basic Pie Chart

## Code

```python
plt.pie(exp_vals)
```

## Explanation

- `plt.pie()` → Pie Chart create karta hai.
- `exp_vals` → Har slice ka size/value determine karta hai.
- Matplotlib values ke proportion ke according slices create karta hai.
- Sabhi slices milkar complete **360°** Pie Chart banate hain.

### Example

Total expense:

```text
1400 + 600 + 300 + 410 + 250 = 2960
```

Home Rent ka percentage:

```text
1400 / 2960 × 100 ≈ 47.3%
```

Isliye Home Rent ka slice sabse bada hoga.

---

# 4. Labels and Percentage

## Code

```python
plt.pie(
    exp_vals,
    labels=exp_labels,
    autopct='%0.1f%%'
)
```

## Explanation

### `labels=exp_labels`

```python
labels=exp_labels
```

- Pie Chart ke slices ke names display karta hai.
- `exp_labels` me diye gaye labels chart par show honge.

### `autopct='%0.1f%%'`

```python
autopct='%0.1f%%'
```

- Pie Chart ke andar percentage display karta hai.
- `%0.1f` → 1 decimal place.
- `%%` → Actual `%` symbol display karta hai.

Example:

```text
47.3%
20.3%
10.1%
13.9%
8.4%
```

## Code

```python
plt.pie(

    autopct='%0.2f%%',
    radius=1.5
)
```

## Explanation

```python
radius=1.5
```

- Pie Chart ka size/radius control karta hai.

- `radius=1.5` → Pie Chart ko larger banata hai.

```text
radius=1   → Normal size
radius=1.5 → Bigger
radius=2   → Aur bigger
```

# 7. `explode`

## Code

```python
plt.pie(
    exp_vals,
    labels=exp_labels,
    autopct='%0.2f%%',
    radius=1.5,
    shadow=True,
    explode=[0, 0.1, 0.1, 0, 0]
)
```

## Explanation

`explode` ka use specific slices ko Pie Chart se **bahar separate/highlight** karne ke liye hota hai.

```python
explode=[0, 0.1, 0.1, 0, 0]
```

Corresponding slices:

```text
Home Rent          → 0
Food               → 0.1
Phone/Internet     → 0.1
Car                → 0
Other Utilities    → 0
```

Isliye:

- Food slice bahar niklega.
- Phone/Internet Bill slice bahar niklega.
- Baaki slices normal position par rahenge.

### Important

`explode` ka order `exp_vals` aur `exp_labels` ke order ke according hota hai.

```text
0    → Normal position
0.1  → Slice thoda bahar
```

---

# 8. `startangle`

## Code

```python
plt.pie(
    exp_vals,
    labels=exp_labels,
    autopct='%0.2f%%',
    radius=1.5,
    shadow=True,
    explode=[0, 0.1, 0.1, 0, 0],
    startangle=45
)
```

## Explanation

```python
startangle=45
```

- Pie Chart ki starting position/rotation set karta hai.
- `45` → Chart ko 45° rotate karke start karega.
- Ye values ya percentages ko change nahi karta.
- Sirf Pie Chart ki **orientation/position** change karta hai.

### Examples

```python
startangle=0
```

→ Default starting position.

```python
startangle=45
```

→ Chart 45° rotate hoga.

```python
startangle=90
```

→ Chart 90° rotate hoga.

# Interview Questions & Answers

### Q1. `plt.pie()` kya karta hai?

**Answer:**

`plt.pie()` numerical values ke basis par **Pie Chart** create karta hai.

---

### Q2. Pie Chart ka main purpose kya hai?

**Answer:**

Total data me different categories ka **proportion/percentage** visualize karna.

---

### Q3. `labels` parameter ka kya use hai?

**Answer:**

Pie Chart ke individual slices/categories ke names display karne ke liye.

---

### Q4. `autopct` kya karta hai?

**Answer:**

Pie Chart ke slices ke andar **percentage values** display karta hai.

---

### Q5. `autopct='%0.1f%%'` ka kya matlab hai?

**Answer:**

Percentage ko **1 decimal place** ke saath display karta hai.

Example:

```text
47.3%
```

---

### Q6. `autopct='%0.2f%%'` ka kya matlab hai?

**Answer:**

Percentage ko **2 decimal places** ke saath display karta hai.

Example:

```text
47.30%
```

---

### Q7. `radius` parameter kya karta hai?

**Answer:**

Pie Chart ka **size/radius** control karta hai.

---

### Q8. `shadow=True` kya karta hai?

**Answer:**

Pie Chart ke neeche shadow effect add karta hai.

---

### Q9. `explode` parameter kya karta hai?

**Answer:**

Specific Pie Chart slice ko baaki slices se **separate/highlight** karne ke liye use hota hai.

---

### Q10. `explode=[0,0.1,0.1,0,0]` ka kya meaning hai?

**Answer:**

```text
0    → Slice normal position
0.1  → Slice thoda bahar
```

Is example me second aur third slices bahar niklenge.

```text
Food               → 0.1
Phone/Internet Bill → 0.1
```

---

### Q11. `startangle` parameter ka use kya hai?

**Answer:**

Pie Chart ki **starting position/rotation** set karne ke liye.

Example:

```python
startangle=45
```

→ Pie Chart 45° rotate hoga.

---

### Q12. Kya `startangle` data ki values change karta hai?

**Answer:**

Nahi.

Ye sirf Pie Chart ki **orientation/position** change karta hai.

---

### Q13. `exp_vals` aur `exp_labels` ka order same hona chahiye?

**Answer:**

Haan.

Example:

```python
exp_vals = [1400, 600]
exp_labels = ["Home Rent", "Food"]
```

Yahan:

```text
1400 → Home Rent
600  → Food
```
